# RAG Context Ablation

Check whether predictions depend on the retrieved context.

```python
from pathlib import Path
import random
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent.parent
sys.path.insert(0, str(ROOT / 'src'))

from structura.dataset import read_jsonl
from structura.inference import load_generator

records = read_jsonl(ROOT / 'data/structura/processed/test.jsonl')
sample = next(record for record in records if record['target']['selected_products'])
generator = load_generator(None)

normal = generator.generate(sample['input']['user_query'], sample['input']['retrieved_context'])['parsed']
empty = generator.generate(sample['input']['user_query'], [])['parsed']
shuffled_context = list(sample['input']['retrieved_context'])
random.shuffle(shuffled_context)
shuffled = generator.generate(sample['input']['user_query'], shuffled_context)['parsed']
{'normal': normal, 'empty': empty, 'shuffled': shuffled}
```
