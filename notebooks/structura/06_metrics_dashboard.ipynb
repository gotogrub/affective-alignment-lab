# Metrics Dashboard

Collect metrics JSON files into a small comparison table.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / 'outputs').exists():
    ROOT = ROOT.parent.parent

rows = []
for path in (ROOT / 'outputs/structura').glob('**/*metrics.json'):
    with path.open('r', encoding='utf-8') as file:
        row = json.load(file)
    row['run'] = path.parent.name
    rows.append(row)
pd.DataFrame(rows)
```
