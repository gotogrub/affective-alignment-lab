# Structura Dataset Preview

Quick checks for JSONL structure, scenario balance and target fields.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent.parent
sys.path.insert(0, str(ROOT / 'src'))

from collections import Counter
from structura.dataset import read_jsonl
from structura.validators import validate_sample_record

path = ROOT / 'data/structura/processed/structura_smoke.jsonl'
records = read_jsonl(path)
len(records), records[0]['id']
```

```python
Counter(record['scenario'] for record in records).most_common()
```

```python
Counter(record['target']['intent'] for record in records).most_common()
```

```python
invalid = [(record['id'], validate_sample_record(record)) for record in records]
[item for item in invalid if item[1]][:10]
```
