# Prompt Format Preview

Inspect how JSONL samples become seq2seq prompts and targets.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent.parent
sys.path.insert(0, str(ROOT / 'src'))

from structura.dataset import read_jsonl
from structura.formatting import format_prompt, format_target

sample = read_jsonl(ROOT / 'data/structura/processed/train.jsonl')[0]
print(format_prompt(sample, template='instruction'))
print('\nTARGET\n', format_target(sample['target']))
```

```python
for template in ['minimal', 'instruction', 'schema-included', 'few-shot']:
    prompt = format_prompt(sample, template=template)
    print(template, len(prompt), prompt[:220].replace('\n', ' '))
```
